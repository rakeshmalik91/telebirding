```python
import json

data = {
    "bird": {},
    "insect": {}
}
with open(f"data/bird-species.json", "r", encoding="utf-8") as file:
    data["bird"]["species"] = json.load(file)
```

```python
for key, details in data["bird"]["species"]['species'].items():
    if 'latin_name' not in details or details['latin_name'].strip() == '':
        print(key)
    if 'ebird_code' not in details or details['ebird_code'].strip() == '':
        print(key)
```

```python
for key, details in data["bird"]["species"]['species'].items():
    if len(details['name'].split(' ')) == 1:
        print(details['name'])
```
